# Modelo de negocio de Vela

Mismo cálculo que la página `/negocio`: el módulo `backend/app/services/business_model.py` lee `data/business/assumptions.csv`. Aquí se puede cambiar un supuesto y ver el resultado sin levantar la aplicación.

```python
import os, sys
sys.path.insert(0, '../backend')
os.environ['BUSINESS_DIR'] = '../data/business'
from app.services import business_model as bm
m = bm.model()
print('Ingreso promedio por institución:', round(m['average_revenue_year1']))
print('Costo fijo mensual año 1:', round(m['fixed_costs_year1']))
print('Punto de equilibrio (instituciones):', m['breakeven_institutions_static'])
print('Mes en que ya no se pierde dinero:', m['breakeven_month'], '· recuperación:', m['payback_month'])
for n in m['npv']:
    print(f"VPN al {n['rate_pct']:.0f} %: {n['npv']:,.0f} MXN")
```

```python
for y in m['years']:
    print(y['year'], round(y['customers_end'], 1), f"{y['revenue']:,.0f}", f"{y['costs']:,.0f}", f"{y['profit']:,.0f}")
```
